## Task1: Use any image processing library and apply the following geometric transformations over the image.

## 1.	Increase the size of the image. 

```python
import cv2

img = cv2.imread('./images/input.jpg')
resized_img = cv2.resize(img, None, fx=1.5, fy=1.5)

print("Original shape:", img.shape)
print("Resized shape:", resized_img.shape)

cv2.imwrite('images/resized_output.jpg', resized_img)
```

## 2.	Rotate the image by 120 degrees.

```python
import cv2

img = cv2.imread('./images/input.jpg')
(h, w) = img.shape[:2]
center = (w // 2, h // 2)

M = cv2.getRotationMatrix2D(center, 120, 1)
rotated_img = cv2.warpAffine(img, M, (w, h))

cv2.imwrite('images/rotated_output.jpg', rotated_img)
print("Done")
```

## 3.	Perform Sheer operations over the image.

```python
import cv2
import numpy as np

img = cv2.imread('./images/input.jpg')
(h, w) = img.shape[:2]

shear_factor = 0.2
M = np.float32([[1, shear_factor, 0],
                 [0, 1, 0]])

sheared_img = cv2.warpAffine(img, M, (w + int(shear_factor * h), h))
cv2.imwrite('images/sheared_output.jpg', sheared_img)
print("Done")
```

## Task2: Use any image processing library and apply the following intensity transformations over the image.

## 1.	Create a Negative Image 

```python
import cv2

img = cv2.imread('./images/input.jpg')
negative_img = 255 - img

cv2.imwrite('images/negative_output.jpg', negative_img)
print("Done")
```

## 2.	Log with (choose an appropriate constant for your image)

```python
import cv2
import numpy as np

img = cv2.imread('./images/input.jpg')

c = 255 / np.log(1 + 255)
log_img = c * np.log(1 + img.astype(np.float64))
log_img = np.array(log_img, dtype=np.uint8)

cv2.imwrite('images/log_output.jpg', log_img)
print("Done")
```

## 3.	Power law (choose a value for gamma that increases the contrast)

```python
import cv2
import numpy as np

img = cv2.imread('./images/input.jpg')

gamma = 2
normalized = img / 255.0
power_img = np.power(normalized, gamma)
power_img = np.array(power_img * 255, dtype=np.uint8)

cv2.imwrite('images/power_output.jpg', power_img)
print("Done")
```
